```python
# Plot histograms from h5 files
import h5py
import numpy as np
import matplotlib.pyplot as plt

def iteratively_loc(labels, h5_file):
    collection = None
    for label in labels:
        if collection is None:
            collection = h5_file[label]
        else:
            collection = collection[label]
    return collection
   
def plot_histograms(h5_file, labels, n_bins=100, x_range=None):
    with h5py.File(h5_file, 'r') as f:
        jet_energy = iteratively_loc(labels, f)
        jet_energy = jet_energy.flatten()

        # Remove nan values
        jet_energy = jet_energy[~np.isnan(jet_energy)]

        # Plot histogram
        if x_range is None:
            plt.hist(jet_energy, bins=n_bins)
        else:
            plt.hist(jet_energy, bins=n_bins, range=x_range)
        plt.xlabel('Jet Energy (GeV)')
        plt.ylabel('Frequency')
        plt.title('Jet Energy Histogram')
        plt.show()

        print('Mean:', np.mean(jet_energy))
        print('Median:', np.median(jet_energy))
        print('Standard Deviation:', np.std(jet_energy))

plot_histograms('../H5Converters/training_run2_odd.h5', ['INPUTS','JET','e'])
plot_histograms('../H5Converters/training_run2_even.h5', ['INPUTS','JET','e'])
plot_histograms('../H5Converters/training_run3_odd.h5', ['INPUTS','JET','e'])
plot_histograms('../H5Converters/training_run3_even.h5', ['INPUTS','JET','e'])
```
